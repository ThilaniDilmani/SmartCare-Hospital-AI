## 2.8 Identify Input Variables

```python
# Display Input Variable
excluded_columns = [
    'record_id',
    'patient_id',
    'no_show',
    'readmitted_30_days',
    'disease_risk_level'
]

input_variables = [
    column for column in df.columns
    if column not in excluded_columns
]

print("Selected target variable:")
print(target_variable)

print("\nNumber of potential input variables:", len(input_variables))

print("\nPotential input variables:")
for variable in input_variables:
    print(variable)
```

## 2.12 Data Quality Findings

```python
print("Duplicate rows:", df.duplicated().sum())
print("\nMissing values per column:")
missing = df.isnull().sum()
print(missing[missing > 0])
print("\nMissing % of room_type:", round(df['room_type'].isnull().mean()*100, 1), "%")
```

The dataset contains 1,000 records and 33 attributes. It includes numerical, categorical, and date-related data. Most columns contain 1,000 non-null values. However, the room_type column contains only 94 non-null values and 906 missing values. This is mainly because a room type is relevant only when a patient is admitted to the hospital. Therefore, these missing values may be structurally expected rather than random data errors.

The selected target variable, disease_risk_level, contains three classes: Low, Medium, and High. The class distribution is not fully balanced because the Low-risk category has fewer records than the Medium- and High-risk categories. This imbalance should be considered during model training and evaluation.

## 2.13 Dataset Overview Summary

The SmartCare Hospital AI dataset contains 1,000 hospital records and 33 attributes. It covers patient demographics, appointments, admissions, clinical measurements, treatments, billing details, and three AI prediction targets.

For this project, `disease_risk_level` was selected as the target variable. It is a multiclass target containing Low, Medium, and High disease-risk categories. Identifier columns and the two alternative AI targets were excluded from the potential input-variable list.

The dataset is generally suitable for machine-learning development because it contains no duplicate records and most attributes have complete values. The main missing-data issue occurs in the `room_type` attribute. Data types, numerical statistics, categorical values, input variables, and the selected target variable were examined during the dataset-understanding process.
